```python
import sqlite3
import datetime
from datetime import timedelta
import pandas as pd

def read_database(db_path, table_name='itin'):
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()

    cursor.execute(f'SELECT * FROM {table_name}')
    rows = cursor.fetchall()
    
    data = pd.DataFrame(rows, columns=[col[0] for col in cursor.description])

    conn.close()
    return data
```

```python
itin_df = read_database('F:\pet-projects\google-flight-scraper\flights.db')
legs_df = read_database('F:\pet-projects\google-flight-scraper\flights.db', table_name='legs')
```

```python
# Check on duplicates
print(itin_df.shape)
print(legs_df.shape)

# Find itinerary_ids in itin_df not in legs_df
itin_only = set(itin_df['itinerary_id']) - set(legs_df['itinerary_id'])
print(f"itinerary_ids in itin_df but not in legs_df: {len(itin_only)}")
if itin_only:
    print(list(itin_only)[:5])  # show a few examples

# Find itinerary_ids in legs_df not in itin_df
legs_only = set(legs_df['itinerary_id']) - set(itin_df['itinerary_id'])
print(f"itinerary_ids in legs_df but not in itin_df: {len(legs_only)}")
if legs_only:
    print(list(legs_only)[:5])  # show a few examples
```

```python
conn = sqlite3.connect('flights copy.db')
cursor = conn.cursor()
cursor.execute("""
    DELETE FROM itin
    WHERE itinerary_id NOT IN (SELECT itinerary_id FROM legs)
""")
conn.commit()
conn.close()
```
